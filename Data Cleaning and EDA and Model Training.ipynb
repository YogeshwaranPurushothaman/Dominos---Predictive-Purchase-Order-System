```python
from statsmodels.tsa.arima.model import ARIMA
import warnings
warnings.filterwarnings("ignore")

df_sale['order_date'] = pd.to_datetime(df_sale['order_date'])

# Group by Date, Pizza_Name, and Pizza_Size to get daily totals
aggregated_df = df_sale.groupby(['order_date', 'pizza_name', 'pizza_size']).agg({'quantity': 'sum'}).reset_index()

# Store trained models for each pizza type and size
trained_models = {}

# Get unique combinations of Pizza_Name and Pizza_Size
pizza_combinations = aggregated_df[['pizza_name', 'pizza_size']].drop_duplicates()

# Training Phase
for _, row in pizza_combinations.iterrows():
    pizza_name = row['pizza_name']
    pizza_size = row['pizza_size']
    
    # Filter data for the specific pizza type and size
    pizza_data = aggregated_df[(aggregated_df['pizza_name'] == pizza_name) & 
                               (aggregated_df['pizza_size'] == pizza_size)]
    
    # Set Date as the index for time series analysis
    pizza_data.set_index('order_date', inplace=True)
    
    # Split data into train and test sets (80% train, 20% test)
    train_size = int(0.8 * len(pizza_data))
    train, test = pizza_data['quantity'][:train_size], pizza_data['quantity'][train_size:]
    
    # Fit ARIMA model (adjust order based on tuning)
    model = ARIMA(train, order=(1, 1, 1))
    model_fit = model.fit()
    
    # Store the trained model
    trained_models[(pizza_name, pizza_size)] = (model_fit, test)

# Create an empty DataFrame to store forecasts and actuals
forecast_results = pd.DataFrame(columns=['Pizza_Name', 'Pizza_Size', 'Forecast_Date', 'Forecast_Quantity', 'Actual_Quantity'])

# Prediction and accuracy calculation
for (pizza_name, pizza_size), (model_fit, test) in trained_models.items():
    # Forecast for the length of the test set
    forecast = model_fit.forecast(steps=len(test))
    
    # Prepare forecast results in DataFrame format
    temp_df = pd.DataFrame({
        'Pizza_Name': pizza_name,
        'Pizza_Size': pizza_size,
        'Forecast_Date': test.index,
        'Forecast_Quantity': np.round(forecast).astype(int),
        'Actual_Quantity': test.values
    })
    
    # Append to forecast_results DataFrame
    forecast_results = pd.concat([forecast_results, temp_df], ignore_index=True)

# Calculate manual accuracy metrics
forecast_results['Absolute_Error'] = np.abs(forecast_results['Actual_Quantity'] - forecast_results['Forecast_Quantity'])
forecast_results['Squared_Error'] = (forecast_results['Actual_Quantity'] - forecast_results['Forecast_Quantity']) ** 2
forecast_results['Percentage_Error'] = np.abs(forecast_results['Absolute_Error'] / forecast_results['Actual_Quantity']) * 100

# Calculate overall metrics
mae = forecast_results['Absolute_Error'].mean()
mse = forecast_results['Squared_Error'].mean()
mape = forecast_results['Percentage_Error'].mean()

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Convert order_date to datetime
df_sale['order_date'] = pd.to_datetime(df_sale['order_date'])

# Group by Date, Pizza_Name, and Pizza_Size to get daily totals
aggregated_df = df_sale.groupby(['order_date', 'pizza_name', 'pizza_size']).agg({'quantity': 'sum'}).reset_index()

# Define forecast steps
forecast_steps = 5

# Store trained SARIMA models for each pizza type and size
trained_models_sarimax = {}

# Get unique combinations of Pizza_Name and Pizza_Size
pizza_combinations = aggregated_df[['pizza_name', 'pizza_size']].drop_duplicates()

# Create an empty DataFrame to store the forecasts and actuals for evaluation
forecast_results = pd.DataFrame(columns=['Pizza_Name', 'Pizza_Size', 'Forecast_Date', 'Forecast_Quantity', 'Actual_Quantity'])

# Training and Evaluation Phase
for _, row in pizza_combinations.iterrows():
    pizza_name = row['pizza_name']
    pizza_size = row['pizza_size']
    
    # Filter data for the specific pizza type and size
    pizza_data = aggregated_df[(aggregated_df['pizza_name'] == pizza_name) & 
                               (aggregated_df['pizza_size'] == pizza_size)]
    
    # Ensure there's enough data
    if len(pizza_data) < 10:  # Adjust based on your needs
        print(f"Not enough data for {pizza_name} - {pizza_size}. Skipping...")
        continue
    
    # Set Date as the index for time series analysis
    pizza_data.set_index('order_date', inplace=True)
    
    # Split data into train and test sets (80% train, 20% test)
    train_size = int(0.8 * len(pizza_data))
    train, test = pizza_data['quantity'][:train_size], pizza_data['quantity'][train_size:]
    
    # Fit SARIMA model
    model = SARIMAX(train, order=(1, 1, 1), seasonal_order=(1, 1, 1, 7))  # Weekly seasonality
    model_fit = model.fit(disp=False)
    
    # Store the trained model
    trained_models_sarimax[(pizza_name, pizza_size)] = model_fit
    
    # Forecast the next steps matching the test set length
    forecast = model_fit.forecast(steps=len(test))
    
    # Prepare forecast results in DataFrame format
    forecast_dates = test.index
    temp_df = pd.DataFrame({
        'Pizza_Name': pizza_name,
        'Pizza_Size': pizza_size,
        'Forecast_Date': forecast_dates,
        'Forecast_Quantity': forecast.values,
        'Actual_Quantity': test.values
    })
    
    # Append the forecast to the main DataFrame
    forecast_results = pd.concat([forecast_results, temp_df], ignore_index=True)

# Calculate manual accuracy metrics
def mean_absolute_error(actual, predicted):
    return np.mean(np.abs(actual - predicted))

def mean_squared_error(actual, predicted):
    return np.mean((actual - predicted) ** 2)

# Calculate errors
mae = mean_absolute_error(forecast_results['Actual_Quantity'], forecast_results['Forecast_Quantity'])
mse = mean_squared_error(forecast_results['Actual_Quantity'], forecast_results['Forecast_Quantity'])
mape = np.mean(np.abs((forecast_results['Actual_Quantity'] - forecast_results['Forecast_Quantity']) / forecast_results['Actual_Quantity'])) * 100

# Display the forecast results and average accuracy metrics
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
```

```python
# Define the forecast date range
from_date = pd.to_datetime(input("Enter the start date (YYYY-MM-DD): "))
to_date = pd.to_datetime(input("Enter the end date (YYYY-MM-DD): "))

# Calculate the number of forecast steps
forecast_steps = (to_date - from_date).days + 1

# Prediction Phase: Create an empty DataFrame to store forecasts
forecast_df_sarimax = pd.DataFrame(columns=['Pizza_Name', 'Pizza_Size', 'Forecast_Date', 'Forecast_Quantity'])

# Assuming trained_models is a dictionary containing your models
# Loop through each trained model for prediction
for (pizza_name, pizza_size), model_fit in trained_models_sarimax.items():
    # Forecast using the trained model
    forecast = model_fit.forecast(steps=forecast_steps)
    
    # Prepare forecast dates and results in DataFrame format
    forecast_dates = pd.date_range(start=from_date, end=to_date)
    temp_df = pd.DataFrame({
        'Pizza_Name': pizza_name,
        'Pizza_Size': pizza_size,
        'Forecast_Date': forecast_dates,
        'Forecast_Quantity': forecast.values
    })
    
    # Append the forecast to the main DataFrame
    forecast_df_sarimax = pd.concat([forecast_df_sarimax, temp_df], ignore_index=True)

# Convert Forecast_Quantity to whole numbers
forecast_df_sarimax['Forecast_Quantity'] = forecast_df_sarimax['Forecast_Quantity'].round(0).astype(int)

# Format the date for the filename
formatted_from_date = from_date.strftime('%Y-%m-%d')  # Format as YYYY-MM-DD
formatted_to_date = to_date.strftime('%Y-%m-%d')      # Format as YYYY-MM-DD

# Merge forecast_df with df_pizza_names_for_map to map pizza_name_id
new_df = forecast_df_sarimax.merge(df_pizza_names_for_map, 
                            left_on=['Pizza_Name', 'Pizza_Size'], 
                            right_on=['pizza_name', 'pizza_size'], 
                            how='left')

# Select only the relevant columns for the new DataFrame
new_df = new_df[['Pizza_Name', 'Pizza_Size', 'Forecast_Date', 'Forecast_Quantity', 'pizza_name_id']]

# Group by pizza_name_id and aggregate Forecast_Quantity
grouped_df = new_df.groupby('pizza_name_id', as_index=False).agg({
    'Pizza_Name': 'first',  # Keep the first pizza name
    'Pizza_Size': 'first',   # Keep the first pizza size
    'Forecast_Quantity': 'sum' # Sum the forecast quantities
})

merged_df = new_df.merge(df_ingredient, 
                          on='pizza_name_id', 
                          how='left')

# Multiply ingredient_qty by Items_Qty_In_Grams
# Assuming Items_Qty_In_Grams is a column in merged_df
merged_df['Total_Ingredient_Qty_In_Gms'] = merged_df['Forecast_Quantity'] * merged_df['Items_Qty_In_Grams']

# Step 3: Group by pizza_ingredients and sum the total quantities
grouped_ingredients_df = merged_df.groupby('pizza_ingredients', as_index=False).agg({
    'Total_Ingredient_Qty_In_Gms': 'sum' # Summing the total quantities for each ingredient
})

# Create the filename without invalid characters
filename_forecast_count_sarimax = f'pizza_sales_sarimax_model_{formatted_from_date}_to_{formatted_to_date}.xlsx'
filename_ingredient_sarimax = f'pizza_ingredients_sarimax_model_{formatted_from_date}_to_{formatted_to_date}.xlsx'

# Export to Excel
forecast_df_sarimax.to_excel(filename_forecast_count_sarimax, index=False)  # Use df.to_excel(filename, index=False) for Excel
grouped_ingredients_df.to_excel(filename_ingredient_sarimax, index=False)

print(f"Data exported to {filename_forecast_count_sarimax}")
print(f"Data exported to {filename_ingredient_sarimax}")
```
